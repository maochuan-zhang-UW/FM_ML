```python
import numpy as np
from scipy.io import loadmat

# Load .mat file
mat = loadmat('W1_CC1_resampled.mat')
data = mat['data']     # shape: (N, 600)
labels = mat['labels'] # shape: (N, 1)

# Reshape to (N, 600, 1) for EQpolarity
data = data.astype(np.float32)
data = data[..., np.newaxis]  # shape: (N, 600, 1)
labels = labels.flatten().astype(int)

# Save as .npy
np.save('your_data.npy', data)
np.save('your_labels.npy', labels)
```
