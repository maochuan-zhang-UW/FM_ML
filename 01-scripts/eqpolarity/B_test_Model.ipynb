```python
import numpy as np
from eqpolarity.utils import construct_model
from sklearn.metrics import accuracy_score

X = np.load('your_data.npy')       # shape: (N, 600, 1)
y_true = np.load('your_labels.npy')  # shape: (N,)

model = construct_model((600,1))
model.load_weights('models/best_weigths_Binary_SCSN_Best.h5')  # Or fine-tuned weights

y_pred_prob = model.predict(X, batch_size=128)
y_pred = (y_pred_prob >= 0.5).astype(int)

acc = accuracy_score(y_true, y_pred)
print(f"Model accuracy on your data: {acc:.2%}")
```
